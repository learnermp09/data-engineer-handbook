```python
# Data Pipeline 1

create_table = spark.sql("CREATE TABLE IF NOT EXISTS dataexpert_students.company_aggregate (company VARCHAR(32), max_age INTEGER, min_age INTEGER)")


aggregated_df = spark.sql("""
    INSERT OVERWRITE TABLE dataexpert_students.company_aggregate
    SELECT
        company,
        MAX(age) AS max_age,
        MIN(age) AS min_age
    FROM dataexpert_students.student_table
    GROUP BY company
""")


df.show()
'''
Displays 20 rows by default.
Does not return the actual rows as Python objects.
Primarily used for inspecting data.
You can specify the number of rows e.g df.show(2)
'''
```
